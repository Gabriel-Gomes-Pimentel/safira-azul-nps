## 3. Pipeline sobre o contrato (card #229)

`criar_pipeline` clona o `ColumnTransformer` do contrato — o ajuste real acontece dentro de cada fold da busca, nunca aqui. As duas células abaixo só ajustam o pipeline sem hiperparâmetros escolhidos e medem a métrica de partida na partição de validação, antes de qualquer busca.

```python
from pipeline_arvore import criar_pipeline, ajustar_no_treino, metrica_de_partida

pipeline_base = criar_pipeline(preparo['preprocessador'])
relato_partida = ajustar_no_treino(pipeline_base, preparo['x']['treino'], y_treino)
relato_partida
```

```python
from avaliacao import avaliar

# Medida na validação, como todas as métricas do protocolo. No treino, a árvore sem limite
# de profundidade decora as linhas e devolveria valores próximos de 1, que não dizem nada
# sobre o desempenho fora da amostra.
metrica_de_partida(pipeline_base, preparo['x']['validacao'], preparo['y']['validacao'], avaliar)
```

## 4. GridSearchCV (card #231)

Usa a grade do card #228 (prefixada pelo nome do passo do pipeline) e os folds agrupados por Cliente de `validacao.criar_folds` — nunca uma partição própria. `scoring=scorer_f2` (card #242, já em `develop`) é o critério único de busca acordado entre as quatro duplas de modelagem (protocolo do card #238) — não mais `average_precision`, que era o placeholder usado enquanto o scorer oficial não existia.

O resultado desta busca, já com `scorer_f2`, está registrado em `documents/extras/busca_arvore_decisao.md` e em `src/hiperparametros_arvore.json` (card #232).

```python
from sklearn.model_selection import GridSearchCV
from validacao import criar_folds
from scorer_f2 import scorer_f2

folds = criar_folds(preparo['x']['treino'], preparo['grupos']['treino'])
grade_prefixada = grade_com_prefixo('modelo')

busca = GridSearchCV(
    criar_pipeline(preparo['preprocessador']),
    grade_prefixada,
    scoring=scorer_f2,
    cv=folds,
    refit=True,
    n_jobs=-1,
)
busca.fit(preparo['x']['treino'], y_treino)

print('best_params_:', busca.best_params_)
print('best_score_:', busca.best_score_)
```

## 5. Hiperparâmetros vencedores (card #232)

Mostra a tabela da busca e confere o vencedor contra `src/hiperparametros_arvore.json`, o registro versionado do card #232. A célula não regrava o arquivo: ele guarda também o critério, as métricas de validação, o número de folhas e a justificativa da escolha, que `best_params_` não tem.

```python
import json
from pathlib import Path

resultados = pd.DataFrame(busca.cv_results_)
colunas_uteis = [c for c in resultados.columns
                 if c.startswith('param_') or c in ('mean_test_score', 'std_test_score', 'rank_test_score')]
tabela_busca = resultados[colunas_uteis].sort_values('rank_test_score')
tabela_busca.head(10)
```

```python
import json

registro = json.loads((RAIZ / 'src' / 'hiperparametros_arvore.json').read_text(encoding='utf-8'))
vencedor = {nome.removeprefix('modelo__'): valor for nome, valor in busca.best_params_.items()}

assert vencedor == registro['hiperparametros'], (
    f'O vencedor da busca {vencedor} difere do registrado em src/hiperparametros_arvore.json '
    f"{registro['hiperparametros']}. Atualize o registro e documents/extras/busca_arvore_decisao.md."
)
print('vencedor confere com src/hiperparametros_arvore.json:', vencedor)
```

## 6. Árvore visual e regras (card #233)

Reconstrói a árvore com os hiperparâmetros vencedores e extrai as regras — esta é a via de explicabilidade intrínseca que o barema pede (ao lado dos odds ratio da Regressão Logística). Mapeie os nomes de saída do `ColumnTransformer` de volta para a feature original antes de interpretar.

```python
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
import matplotlib.pyplot as plt

# best_estimator_ é o pipeline vencedor já reajustado no treino inteiro (refit=True).
modelo_final = busca.best_estimator_

arvore = modelo_final.named_steps['modelo']
nomes_features = modelo_final.named_steps['preparo'].get_feature_names_out()

fig, ax = plt.subplots(figsize=(20, 10))
plot_tree(arvore, feature_names=nomes_features, class_names=['nao Detrator', 'Detrator'],
          filled=True, rounded=True, fontsize=8, ax=ax)
plt.show()
```

```python
regras = export_text(arvore, feature_names=list(nomes_features))
print(regras)
```
