# Definição da unidade de predição

Este notebook responde, antes de qualquer modelagem, a duas perguntas sobre a granularidade da base analítica (`data/processed/base_analitica.parquet`, gerada pela seção 13 do `pre-processamento.ipynb`):

1. **Cliente ou resposta?** As duas células de `groupby` contam quantas respostas (`RESPONDENT_ID`) cada Cliente (`ID_GOLDENRECORD`) tem. A média é de 1,19 resposta por Cliente, com mediana de 1, e por isso a unidade de predição é a resposta. A recorrência que existe é tratada na divisão por Cliente do `modelagem.ipynb`.
2. **Jornada ou trecho?** As duas últimas células procuram colunas de rota e contam os aeroportos de `BASE_AIRPORTLEG`. A jornada permanece uma linha, e o número de trechos vira a feature `N_TRECHOS`.

A leitura dos resultados está na Seção 4.3.2.2 de `documents/documentacao.md`.
